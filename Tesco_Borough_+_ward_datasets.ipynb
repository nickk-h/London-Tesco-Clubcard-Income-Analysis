```python
product_cols = [
    'f_beer','f_dairy','f_eggs','f_fats_oils','f_fish',
    'f_fruit_veg','f_grains','f_meat_red','f_poultry',
    'f_readymade','f_sauces','f_soft_drinks',
    'f_spirits','f_sweets','f_tea_coffee',
    'f_water','f_wine'
]


nutrient_cols = [
    'fat','saturate','salt','sugar',
    'protein','carb','fibre','alcohol',
    'energy_tot','h_nutrients_calories'
]
```

```python
def seasonal_products(df):
  monthly_products = df.groupby("month")[product_cols].mean()
  seasonality_products = (monthly_products.std() / monthly_products.mean()).sort_values(ascending=False)
  return monthly_products,seasonality_products
```

```python
def plot_seasonality(monthly_products, seasonality_products, monthly_std=None):

    import seaborn as sns
    import calendar


    sns.set_theme(style="white")

    top5 = seasonality_products.head(5).index
    data = monthly_products[top5]

    label_map = {
        "f_spirits":    "Spirits",
        "f_beer":       "Beer",
        "f_wine":       "Wine",
        "f_tea_coffee": "Tea & Coffee",
        "f_sauces":     "Sauces",
    }

    months = np.arange(1, 13)
    month_labels = [calendar.month_abbr[i] for i in months]
    colors = ["#2E4057", "#3A7CA5", "#5F8F8F", "#7A9E7E", "#6C757D"]

    swings = {col: data[col].max() - data[col].min() for col in top5}

    fig, ax = plt.subplots(figsize=(12, 6.5))

    for i, col in enumerate(top5):
        color = colors[i]
        values = data[col]

        if monthly_std is not None and col in monthly_std.columns:
            std_vals = monthly_std[col].values
            ax.fill_between(
                months,
                values.values - std_vals,
                values.values + std_vals,
                color=color, alpha=0.10, linewidth=0,
            )


        ax.plot(
            months, values,
            marker="o", linewidth=2.5, markersize=4,
            color=color, label=label_map.get(col, col), zorder=3,
        )


        ax.axhline(
            values.mean(),
            color=color, linestyle=":", linewidth=0.8, alpha=0.45, zorder=1,
        )


        max_month = values.idxmax()
        max_value = values.max()
        ax.scatter(max_month, max_value, marker="o", s=100,
                   color=color, edgecolors="white", linewidth=1.5, zorder=6)
        ax.text(max_month, max_value * 1.022,
                calendar.month_abbr[max_month],
                ha="center", fontsize=8.5, color=color, fontweight="bold")


        min_month = values.idxmin()
        min_value = values.min()
        ax.scatter(min_month, min_value, marker="o", s=100,
                   facecolors="white", edgecolors=color, linewidth=2, zorder=6)
        ax.text(min_month, min_value * 0.964,
                calendar.month_abbr[min_month],
                ha="center", va="top", fontsize=8.5, color=color)


        swing_pct = (max_value - min_value) / min_value * 100
        top2_cols = sorted(swings, key=swings.get, reverse=True)[:2]
        if col in top2_cols:
            ax.annotate(
                f"+{swing_pct:.0f}% swing",
                xy=(max_month, max_value),
                xytext=(max_month + 0.35, max_value * 1.04),
                fontsize=7.5, color=color, alpha=0.85,
                arrowprops=dict(arrowstyle="-", color=color, alpha=0.4, lw=0.8),
            )


    ax.set_title(
        "Seasonal Consumption Patterns of Selected Grocery Categories",
        fontsize=14, weight="bold", pad=14,
    )
    ax.text(
        0.5, 1.012,
        "Dotted line = category annual mean  ·  ● peak  ○ trough"
        + ("  ·  Shaded band = ±1 SD" if monthly_std is not None else ""),
        transform=ax.transAxes, ha="center", fontsize=8, color="#666666",
    )

    ax.set_xlabel("Month", labelpad=8)
    ax.set_ylabel("Average Consumption", labelpad=8)
    ax.set_xticks(months)
    ax.set_xticklabels(month_labels)


    handles, labels_l = ax.get_legend_handles_labels()
    label_to_col = {v: k for k, v in label_map.items()}
    sorted_pairs = sorted(
        zip(labels_l, handles),
        key=lambda x: swings.get(label_to_col.get(x[0], ""), 0),
        reverse=True,
    )
    sorted_labels, sorted_handles = zip(*sorted_pairs)

    leg = ax.legend(
        sorted_handles, sorted_labels,
        frameon=False,
        bbox_to_anchor=(1.02, 1), loc="upper left",
        title="Category  (↓ by swing)", title_fontsize=8.5, fontsize=9,
    )
    for text in leg.get_texts():
        col_key = label_to_col.get(text.get_text())
        if col_key and col_key in swings:
            pct = swings[col_key] / data[col_key].min() * 100
            text.set_text(f"{text.get_text()}  (+{pct:.0f}%)")

    sns.despine()
    plt.tight_layout()
    plt.show()
```

```python
def plot_seasonality_clean(monthly_products, seasonality_products):


    import calendar


    top5 = seasonality_products.head(5).index
    data = monthly_products[top5]

    label_map = {
        "f_spirits":    "Spirits",
        "f_beer":       "Beer",
        "f_wine":       "Wine",
        "f_tea_coffee": "Tea & Coffee",
        "f_sauces":     "Sauces",
    }

    colors = {
        "f_spirits": "#4CC9F0",
        "f_beer": "#FF6B6B",
        "f_wine": "#F72585",
        "f_tea_coffee": "#2EC4B6",
        "f_sauces": "#4361EE",
    }

    months = np.arange(1, 13)
    month_labels = [calendar.month_abbr[i] for i in months]


    fig = plt.figure(figsize=(13.33, 7.5), facecolor='none')
    ax = plt.axes([0.06, 0.12, 0.88, 0.75])
    ax.set_facecolor('none')

    offsets = np.linspace(0, 0.8, len(top5))


    for i, col in enumerate(top5):
        values = data[col]
        norm = values / values.mean()
        color = colors[col]
        norm = norm + offsets[i]

        ax.plot(months, norm, color=color, linewidth=2.8)


        max_m = values.idxmax()
        max_v = norm.iloc[max_m - 1]
        ax.scatter(max_m, max_v, s=70, color=color,
                   edgecolors="black", linewidth=1.5, zorder=5)
        ax.vlines(x=max_m, ymin=max_v, ymax=norm.max() + 0.6,
                  color=color, linewidth=2.5, linestyle='dashed', alpha=0.8)


        min_m = values.idxmin()
        min_v = norm.iloc[min_m - 1]
        ax.scatter(min_m, min_v, s=70,
                   facecolors="white", edgecolors=color,
                   linewidth=2, zorder=5)
        ax.vlines(x=min_m, ymin=norm.min() - 0.6, ymax=min_v,
                  color=color, linewidth=2.5, linestyle='dashed', alpha=0.8)


        ax.text(12.3, norm.iloc[-1],
                label_map[col],
                fontsize=11,
                color=color,
                va="center",
                fontweight='bold')


    season_lines = [3, 6, 9, 12]
    for x in season_lines:
        ax.axvline(x=x - 0.5, color='#B0B0B0', linestyle='--', linewidth=1)


    season_positions = [1.5, 4.5, 7.5, 10.5]
    season_names = ['Winter', 'Spring', 'Summer', 'Autumn']

    y_top = ax.get_ylim()[1]

    for pos, name in zip(season_positions, season_names):
        ax.text(pos, y_top + 0.1,
                name,
                ha='center',
                fontsize=11,
                color='black',
                fontweight='bold')


    ax.set_xticks(months)
    ax.set_xticklabels(month_labels, color='black', fontsize=10)
    ax.set_yticks([])


    for spine in ax.spines.values():
        spine.set_visible(False)

    plt.show()
```

```python

import statsmodels.api as sm


df = df_msoa_year_copy.copy()


df["Average_Income"] = (
    df["Average_Income"]
    .astype(str)
    .str.replace(",", "")
    .astype(float)
)


df = df[df["Average_Income"] > 0]


df["log_income"] = np.log(df["Average_Income"])



def run_reg(y):
    X = df[["log_income", "avg_age", "people_per_sq_km", "population"]]
    X = sm.add_constant(X)

    model = sm.WLS(
        df[y],
        X,
        weights=df["population"],
        missing='drop'
    ).fit()

    print("\n=========================")
    print(f"RESULTS: {y}")
    print("=========================")
    print(model.summary())

    return model



m_dairy = run_reg("f_dairy")
m_fruitveg = run_reg("f_fruit_veg")
m_fats = run_reg("f_fats_oils")
m_sweets = run_reg("f_sweets")



m_ready = run_reg("f_readymade")



X = df[["log_income", "f_readymade", "avg_age", "people_per_sq_km", "population"]]
X = sm.add_constant(X)

m_fresh = sm.WLS(
    df["f_fruit_veg"],
    X,
    weights=df["population"],
    missing='drop'
).fit()

print("\n=========================")
print("RESULTS: Fresh food vs Ready meals")
print("=========================")
print(m_fresh.summary())



print("\n\n===== KEY COEFFICIENTS =====")

print("\nIncome effects:")
print("Dairy:", m_dairy.params["log_income"])
print("Fruit & Veg:", m_fruitveg.params["log_income"])
print("Fats:", m_fats.params["log_income"])
print("Sweets:", m_sweets.params["log_income"])
print("Ready meals:", m_ready.params["log_income"])

print("\nReady meals effect on fresh food:")
print("f_readymade → f_fruit_veg:", m_fresh.params["f_readymade"])
```

```python
from scipy.stats import spearmanr

numeric_cols = df.columns.drop(['area_id','MSOA'])
df[numeric_cols] = df[numeric_cols].apply(pd.to_numeric, errors='coerce')

df['deprivation_rank'] = df['Average_Income'].rank(ascending=True)

nutrient_indicators = [
    'fibre', 'sugar', 'fat', 'saturate',
    'energy_tot', 'energy_fat', 'energy_sugar'
]

food_categories = [
    'f_fruit_veg', 'f_sweets', 'f_soft_drinks'
]

def spearman_results(df, cols):
    results = []
    for col in cols:
        corr, p = spearmanr(df['deprivation_rank'], df[col], nan_policy='omit')
        results.append((col, corr, p))
    return pd.DataFrame(results, columns=['variable','spearman_corr','p_value'])

nutrient_corr = spearman_results(df, nutrient_indicators)
food_corr = spearman_results(df, food_categories)

print("Spearman Correlations with Deprivation (Lower Income = Higher Rank):")
print(nutrient_corr)
print(food_corr)

sns.set(style="whitegrid", context="talk")
df = df.copy()

all_cols = nutrient_indicators + food_categories

for i, col in enumerate(all_cols):
    fig, ax = plt.subplots(figsize=(6, 4))
    sns.scatterplot(x='Average_Income', y=col, data=df, ax=ax)
    sns.regplot(x='Average_Income', y=col, data=df,
                scatter=False, color='red', ax=ax)
    ax.set_title(f"{col} vs Average Income", pad=12)
    ax.set_xlabel("Average Income (£)", labelpad=10)
    ax.set_ylabel(col, labelpad=10)
    fig.tight_layout(pad=2.5)
    plt.subplots_adjust(top=0.88)
    plt.show()
    plt.close(fig)
    print("\n" * 3)

top_quant = df['Average_Income'].quantile(0.8)
bot_quant = df['Average_Income'].quantile(0.2)

df_top = df[df['Average_Income'] >= top_quant]
df_bot = df[df['Average_Income'] <= bot_quant]

def compare_means(df_high, df_low, cols):
    summary = []
    for c in cols:
        h_mean = df_high[c].mean()
        l_mean = df_low[c].mean()
        diff = h_mean - l_mean
        summary.append((c, h_mean, l_mean, diff))
    return pd.DataFrame(summary, columns=['variable','top20_mean','bottom20_mean','difference'])

nutrient_comp = compare_means(df_top, df_bot, nutrient_indicators)
food_comp     = compare_means(df_top, df_bot, food_categories)

print("Top vs Bottom 20% Income Areas - Nutrient Comparison:")
print(nutrient_comp)
print("Top vs Bottom 20% Income Areas - Food Categories:")
print(food_comp)
```

```python
from scipy.stats import spearmanr

sns.set_theme(style="whitegrid")

analysis_columns = [
    'name',
    'Average_Income',
    'f_fruit_veg',
    'f_fish',
    'f_soft_drinks',
    'f_sweets',
    'f_readymade',
    'f_fats_oils',
    'f_sauces',
    'f_water',
    'f_wine',
    'f_beer',
    'f_spirits',
    'num_transactions'
]

df = merged_gdf[analysis_columns].copy()
df = df.rename(columns={'name': 'borough'})
df = df[df['Average_Income'] > 0]


df['healthy_index'] = (
    df['f_fruit_veg'] +
    df['f_fish'] +
    df['f_water']
)

df['unhealthy_index'] = (
    df['f_sweets'] +
    df['f_soft_drinks'] +
    df['f_readymade'] +
    df['f_fats_oils']
)

df['net_healthiness'] = df['healthy_index'] - df['unhealthy_index']


metrics = [
    'f_fruit_veg',
    'f_fish',
    'f_soft_drinks',
    'f_sweets',
    'f_readymade',
    'f_fats_oils',
    'healthy_index',
    'unhealthy_index',
    'net_healthiness',
    'num_transactions'
]


results = []

for metric in metrics:

    rho, p = spearmanr(df['Average_Income'], df[metric])

    results.append({
        "metric": metric,
        "spearman_rho": rho,
        "p_value": p
    })

corr_df = pd.DataFrame(results)

corr_df = corr_df.sort_values(
    by="spearman_rho",
    ascending=True
)



label_map = {
    'f_fruit_veg': 'Fruit & Vegetables',
    'f_fish': 'Fish',
    'f_soft_drinks': 'Soft Drinks',
    'f_sweets': 'Sweets',
    'f_readymade': 'Ready Meals',
    'f_fats_oils': 'Fats & Oils',
    'healthy_index': 'Healthy Basket Index',
    'unhealthy_index': 'Unhealthy Basket Index',
    'net_healthiness': 'Net Healthiness',
    'num_transactions': 'Number of Transactions'
}

corr_df['metric_label'] = corr_df['metric'].map(label_map)


plt.figure(figsize=(10,6))

ax = sns.barplot(
    data=corr_df,
    x="spearman_rho",
    y="metric_label",
    palette="viridis"
)

plt.axvline(0, color='black', linewidth=1)

plt.title(
    "Relationship Between Income and Food Purchasing Behaviour\n"
    "(Spearman Correlation)"
)

plt.xlabel("Spearman Correlation with Average Income")
plt.ylabel("Food Category / Behaviour")

for i, v in enumerate(corr_df['spearman_rho']):
    ax.text(v + 0.01 if v > 0 else v - 0.05, i,
            f"{v:.2f}", va='center')

plt.tight_layout()
plt.show()
```
